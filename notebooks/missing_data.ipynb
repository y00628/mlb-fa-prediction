```python
import glob
import pandas as pd
```

```python
csv_files = glob.glob('../data/raw/*.csv')
data = pd.DataFrame()
for file in csv_files:
    temp = pd.read_csv(file)
    temp['year'] = file.split('/')[-1].split('.')[0][:4]
    data = pd.concat([data, temp], ignore_index=True)
```

```python
data.to_csv('../data/processed/data.csv', index=False)
```
